```python
nama_siswa = input("Masukkan Nama Siswa: ")
jam_datang = float(input("Masukkan Jam Datang (contoh 6.50 atau 7.10): "))

if jam_datang <= 7.00:
    status = "Hadir Tepat Waktu"
elif jam_datang <= 7.15:
    status = "Hadir Terlambat (Toleransi)"
else:
    status = "Terlambat Parah (Lapor Guru Piket)"

print("--------------------------------")
print(f"Siswa : {nama_siswa}")
print(f"Jam   : {jam_datang}")
print(f"Status: {status}")
print("--------------------------------")
```
