```python
%%writefile kelilingpersegipanjang.cpp
#include <iostream>
using namespace std;

int main()
{
    int p, l, k;

    cout << "Masukkan panjang (P): ";
    cin >> p;

    cout << "Masukkan lebar (L): ";
    cin >> l;

    k = 2 * (p + l);

    cout << "Hasil keliling persegi panjang = " << k << endl;

    return 0;
}
```
