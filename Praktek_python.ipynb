```python
import ipywidgets as widgets
from IPython.display import display

# Membuat input nama pemain
input_nama = widgets.Text(
    value="",
    placeholder='ketik nama pemain',
    description='Nama:',
    disabled=False
)

# Membuat pilihan level menggunakan slider
input_level = widgets.IntSlider(
    min=1,
    max=100,
    description='Level:'
)

# Membuat pilihan role menggunakan dropdown
input_role = widgets.Dropdown(
    options=['Warrior', 'Mage', 'Archer'],
    description='Role:'
)

# Membuat tombol daftar
tombol_daftar = widgets.Button(
    description='Daftar',
    button_style='primary'
)

# Tempat menampilkan hasil
output_hasil = widgets.Output()

# Membuat fungsi saat tombol diklik
def daftar_klik(b):
    with output_hasil:
        output_hasil.clear_output()
        print(
            f"Pemain {input_nama.value} dengan role "
            f"{input_role.value} level {input_level.value} "
            "telah siap bertarung!"
        )

# Menghubungkan tombol dengan fungsi
tombol_daftar.on_click(daftar_klik)

# Menampilkan semua widget
display(
    input_nama,
    input_level,
    input_role,
    tombol_daftar,
    output_hasil
)
```
